```python
import numpy as np

rng = np.random.default_rng(42)

marks = rng.integers(30, 101, size=(100, 5))

print("Marks:\n", marks)
print("Shape:", marks.shape)

per_student = marks.mean(axis=1)
per_exam = marks.mean(axis=0)

print("Student averages:\n", per_student)
print("Exam averages:\n", per_exam)

passed = per_student >= 60

passed_count = passed.sum()
passed_fraction = passed.mean()

print("Students passed:", passed_count)
print("Fraction passed:", passed_fraction)

top_10_indices = np.argsort(per_student)[-10:][::-1]

top_10_scores = per_student[top_10_indices]

print("Top 10 student indices:", top_10_indices)
print("Top 10 student averages:", top_10_scores)

exam_mean = marks.mean(axis=0)
exam_std = marks.std(axis=0)

standardised = (marks - exam_mean) / exam_std

print("Standardised marks:\n", standardised)

print("New exam means:", standardised.mean(axis=0))
print("New exam standard deviations:", standardised.std(axis=0))

curved_marks = np.minimum(marks + 5, 100)

print("Curved marks:\n", curved_marks)

capped_count = (curved_marks == 100).sum()

print("Marks that reached the cap:", capped_count)

failed = marks < 40

failed_every_exam = np.all(failed, axis=1)

failed_students = np.where(failed_every_exam)[0]

print("Students who failed every exam:", failed_students)

```
