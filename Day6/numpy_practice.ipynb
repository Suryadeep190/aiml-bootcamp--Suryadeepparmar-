```python
import numpy as np

def python_square(data):
    result = []

    for value in data:
        result.append(value ** 2)

    return result


def numpy_square(arr):
    return arr ** 2


def python_sum(data):
    total = 0

    for value in data:
        total += value

    return total


def numpy_sum(arr):
    return np.sum(arr)


def python_dot(a, b):
    total = 0

    for i in range(len(a)):
        total += a[i] * b[i]

    return total


def numpy_dot(a, b):
    return np.dot(a, b)

    
data = [2, 3, 4]
a = [2, 3, 4]
b = [5, 6, 7]

print("Python square:", python_square(data))
print("NumPy square:", numpy_square(np.array(data)))

print("Python sum:", python_sum(data))
print("NumPy sum:", numpy_sum(np.array(data)))

print("Python dot product:", python_dot(a, b))
print("NumPy dot product:", numpy_dot(np.array(a), np.array(b)))
```

```python
import timeit

sizes = [1_000, 10_000, 100_000, 1_000_000]

results = []

for size in sizes:

    # Generate data outside the timer
    data = np.random.randint(1, 100, size=size).tolist()
    arr = np.array(data)

    a = data
    b = np.random.randint(1, 100, size=size).tolist()

    a_np = np.array(a)
    b_np = np.array(b)

    # Measure Python operations
    python_square_time = timeit.timeit(
        lambda: python_square(data), number=10
    )

    python_sum_time = timeit.timeit(
        lambda: python_sum(data), number=10
    )

    python_dot_time = timeit.timeit(
        lambda: python_dot(a, b), number=10
    )

    # Measure NumPy operations
    numpy_square_time = timeit.timeit(
        lambda: numpy_square(arr), number=10
    )

    numpy_sum_time = timeit.timeit(
        lambda: numpy_sum(arr), number=10
    )

    numpy_dot_time = timeit.timeit(
        lambda: numpy_dot(a_np, b_np), number=10
    )

    # Calculate speedup
    results.append({
        "Size": size,
        "Square Speedup": python_square_time / numpy_square_time,
        "Sum Speedup": python_sum_time / numpy_sum_time,
        "Dot Speedup": python_dot_time / numpy_dot_time
    })


# Display results
print(f"{'Size':<12}{'Square':<15}{'Sum':<15}{'Dot Product':<15}")

for row in results:
    print(
        f"{row['Size']:<12,}"
        f"{row['Square Speedup']:.2f}x{'':<11}"
        f"{row['Sum Speedup']:.2f}x{'':<11}"
        f"{row['Dot Speedup']:.2f}x"
    )
```

```python
import matplotlib.pyplot as plt

for operation in ["Square Speedup", "Sum Speedup", "Dot Speedup"]:
    plt.plot(
        sizes,
        [row[operation] for row in results],
        marker="o",
        label=operation
    )

plt.xscale("log")

plt.xlabel("Array Size")
plt.ylabel("Speedup (Python / NumPy)")
plt.title("Python Loops vs NumPy Performance")

plt.legend()
plt.grid(True)
plt.show()

#Numpy may not have advantage in speed while working with small arrays because setting up operationscan take time relative to small work performed.
```
