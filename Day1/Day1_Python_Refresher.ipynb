```python
def reverse_words(sentence):
    words = sentence.split()
    reversed_words= words[::-1]
    return ' '.join(reversed_words)

print(reverse_words("hello there I am Ai/Ml intern"))
```

```python
def count_words(sentence):
    count={}
    for word in sentence.lower().split():
        count[word]=count.get(word,0)+1
    return count

print(count_words("there is a cat wearing  a hat "))
```

```python
def is_palindrome(word):
    clean_word=word.lower()
    return clean_word==clean_word[::-1]

print(is_palindrome("Racecar"))
print(is_palindrome("Suryadeep"))
```

```python
def calculate_stats(numbers: list[float]) -> dict[str, float | int | None]:
  if not numbers:
    return {
        "min": None,
        "max": None,
        "mean": None,
        "count": 0,
    }

  count = len(numbers)
  return {
      "min": min(numbers),
      "max": max(numbers),
      "mean": sum(numbers) / count,
      "count": count,
  }


print(calculate_stats([10, 20, 30, 40]))

print(calculate_stats([]))
```

```python
def safe_divide(a: float, b: float) -> float | str:
    try:
        return a/b
    except ZeroDivisionError:
        return "Error: no number can be divided by zero"
    
print(safe_divide(10, 2))
print(safe_divide(11, 0))    
```

```python
import time

def find_primes(n: int) -> list[int]:
    primes = []
    for num in range(2, n + 1):
        is_prime = True
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            primes.append(num)
    return primes

start_time = time.perf_counter()
primes_list=find_primes(10000)
end_time=time.perf_counter()

print(f"found{len(primes_list)}primes upto 10000 ")
print(f"Time taken: {(end_time - start_time) * 1000:.3f} ms")
```

```python
from collections import Counter

def count_words_counter(sentence: str) -> Counter:
    return Counter(sentence.lower().split())

text = "This is a international cricket match and a final too"
print(count_words_counter(text))    
```

```python
import random


def play_guessing_game() -> None:
  secret = random.randint(1, 100)
  attempts = 0

  print("I'm thinking of a number between 1 and 100.")

  while True:
    guess_text = input("Enter your guess: ")
    if not guess_text.isdigit():
      print("Please enter a valid whole number.")
      continue

    guess = int(guess_text)
    attempts += 1

    if guess < secret:
      print("Higher!")
    elif guess > secret:
      print("Lower!")
    else:
      print(f"Correct! You got it in {attempts} tries.")
      break


play_guessing_game()
```
